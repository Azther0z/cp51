# TO DO 5: หาค่ามัธยฐานและค่าเฉลี่ยของแต่ละวิชา และนำไปแสดงผลบนกราฟแท่งแบบ Multiple plot

```python
#Write code below please.
mean_math = csv_data['math score'].mean()
mean_writing = csv_data['writing score'].mean()
mean_reading = csv_data['reading score'].mean()

median_math = csv_data['math score'].median()
median_reading = csv_data['reading score'].median()
median_writing = csv_data['writing score'].median()
```

```python
print(mean_math)
```

```python
#Plot multiple bar graph
#Write code below please.
N = 3
mean = (mean_math, mean_reading, mean_writing)
median = (median_math, median_reading, median_writing)
width = 0.40 
index_mean = [idx for idx in range(N)]
index_median = [idx+0.40 for idx in range(N)]
index_number = [idx+0.40 / 2 for idx in range(N)]
plt.bar(index_mean,mean, width, label='Mean Score',color='pink')
plt.bar(index_median,median, width, label='Median Score',color='cyan')
plt.ylabel('Scores')
plt.title('Scores by group and gender')
plt.xticks(index_number, ('Math', 'Reading', 'Writing',))
plt.legend(loc='best')
plt.show()
```
