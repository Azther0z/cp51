## Problem 2: Sound Denoising with FFT
In this problem, a synthesized multi-tone signal is corrupted by heavy additive Gaussian and random noise. You are required to analyze the frequency spectrum using FFT and design a frequency-domain threshold filter.

```python
import numpy as np
import pandas as pd
from scipy import signal,fftpack
import cv2
from skimage.io import imread
import matplotlib.pyplot as plt
import IPython.display as ipd
import os

import librosa
import parselmouth
```

```python
sampling_rate = 32000
N=10001
Nf = 3 # Nf--> num freq
t= np.arange(N,dtype=float)
# pick rand period betwwen 10-2010 and convert to freq

# random period
Ts = np.random.rand(Nf)*2000+10
fs=1/Ts

# fs in sampling rate = 32000
fs_real = fs*sampling_rate

# pick rand Amp and phase
amp = np.random.rand(Nf)*200+ 100
phi = np.random.rand(Nf)*2*np.pi

# create clean signal
h = np.zeros(N)
for i in range(len(fs)):
    h += amp[i]*np.sin(2*np.pi*fs[i]*t + phi[i])

# signal with noise
h_w_noise = h + np.random.randn(N)*3*h + np.random.randn(N)*700
```

```python
# TODO 2.1 : plot (1) clean signal and (2) noisy signal with label
plt.plot(t, h, label='h')
plt.plot(t, h_w_noise, label='h_w_noise',alpha=0.2)
plt.legend()
plt.xlabel('Time')
plt.ylabel('Signal Value')
plt.show()
```

```python
# TODO 2.2: plot magnitude of FFT of the noisy signal (freq sort form min--> max)
fftH = np.fft.fft(h_w_noise)
fftFreq = np.fft.fftfreq(N,d=1/sampling_rate)
fftMag = np.abs(fftH)/N

order = np.argsort(fftFreq)

plt.plot(fftFreq[order],fftMag[order])
plt.xlabel('Frequency (Hz)')
plt.ylabel('FFT Magnitude')
plt.show()
```

```python
# TODO 2.3 : cleaning the noisy signal using magnitude of FFT
fftHClean = fftH.copy()
fftHClean[fftMag < 30] = 0
```

```python
# TODO 2.4 : plot clean signal, noise signal and filtered signal (from your result in TODO 2.3) with label

h_clean = np.fft.ifft(fftHClean).real

plt.plot(t, h_w_noise, alpha=0.2, label="h_w_noise")
plt.plot(t, h_clean, label="h_clean")
plt.plot(t, h, label="h")
plt.xlabel('Time')
plt.ylabel('Signal Value')
plt.legend()
plt.show()
```

```python
# TODO 2.5 : export with IPython.display, listen to (1) original signal (2) signal with noise (3) signal after filtered

ipd.display(ipd.Audio(h,rate=sampling_rate))
ipd.display(ipd.Audio(h_w_noise,rate=sampling_rate))
ipd.display(ipd.Audio(h_clean,rate=sampling_rate))
```
